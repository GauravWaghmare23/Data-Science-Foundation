```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 5, 11)
x
```

```python
y = x**2
y
```

```python
plt.plot(x, y)
plt.title("Square Numbers Line Graph")
plt.xlabel("Value")
plt.ylabel("Square of Value")
```

```python
x = [1, 2, 3, 4, 5]
y = [1, 4, 9, 16, 25]
```

```python
plt.figure(figsize=(12, 8)) 

# 1. Line Plot
plt.subplot(2, 2, 1)
plt.plot(x, y)
plt.title("Square Numbers Line Graph")
plt.xlabel("Value")
plt.ylabel("Square of Value")

# 2. Scatter Plot
plt.subplot(2, 2, 2)
plt.scatter(x, y)
plt.title("Square Numbers Scatter Plot")
plt.xlabel("Value")
plt.ylabel("Square of Value")

# 3. Bar Chart
plt.subplot(2, 2, 3)
plt.bar(x, y)
plt.title("Square Numbers Bar Chart")
plt.xlabel("Value")
plt.ylabel("Square of Value")

# 4. Histogram
plt.subplot(2, 2, 4)
plt.hist(y, bins=5)
plt.title("Square Numbers Histogram")
plt.xlabel("Square Value")
plt.ylabel("Frequency")

plt.tight_layout(pad=3.0)
```

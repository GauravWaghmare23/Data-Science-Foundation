```python
import numpy as np
```

Numpy Indexing and slicing of vectors

```python
arr = np.arange(1,11)
```

```python
print(arr)
```

```python
print(arr[9])
```

```python
print(arr[2:7])
print(arr[:5])
print(arr[5:])
print(arr[:])
```

Matix Indexing and slicing

```python
arr = np.arange(1,31).reshape(5,6)
print(arr)
```

```python
print(arr[0:2,2:4])
```

```python
print(arr[2:,4:])
```

```python
print(arr[:,3])
```

```python
print(arr[2,:])
```

Boolean Indexing

```python
print(arr)
```

```python
bool_index = arr % 2 == 0
print(bool_index)
```

```python
arr_even = arr[bool_index]
print(arr_even)
```

Matrix Operations

```python
mat1 = np.random.randint(1, 11, (3, 3))
mat2 = np.random.randint(1, 11, (3, 3))
```

```python
print(mat1)
```

```python
print(mat2)
```

```python
mat_multiplication = mat1 @ mat2
print(mat_multiplication)
```

```python
mat_multiplication = np.dot(mat1, mat2)
print(mat_multiplication)
```

```python
transpose = mat_multiplication.T
print(transpose)
```

Advance Array Manipulations

Array Stacking

```python
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])
arr3 = np.array([7, 8, 9])
```

```python
vertical_stacked = np.vstack((arr1, arr2, arr3))
print(vertical_stacked)
```

```python
horizontal_stacked = np.hstack((arr1, arr2, arr3))
print(horizontal_stacked)
```

```python
stacked = np.stack((arr1, arr2, arr3), axis=0)
print(stacked)
```

```python
stacked = np.stack((arr1, arr2, arr3), axis=1)
print(stacked)
```

```python
stacked = np.stack((arr1, arr2), axis=1)
print(stacked)
```

```python
mat = np.random.randint(1, 11, (3, 3))
print(mat)
```

```python
vertical_split = np.vsplit(mat, 3)
print(vertical_split)
```

```python
horizontal_split = np.hsplit(mat, 3)
print(horizontal_split)
```

```python
import pandas as pd
import numpy as np

data = {
    'Date': pd.date_range('2023-01-01', periods=120),

    'Product': (
        ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H'] * 15
    ),

    'Region': (
        [
            'East', 'West', 'North', 'South',
            'Central', 'Northeast', 'Northwest', 'Southeast'
        ] * 15
    ),

    'Sales': np.random.randint(100, 5000, 120),

    'Units': np.random.randint(10, 500, 120),

    'Rep': (
        [
            'John', 'Mary', 'Bob', 'Alice',
            'David', 'Emma', 'Michael', 'Sophia',
            'James', 'Olivia', 'Daniel', 'Charlotte'
        ] * 10
    )
}

df = pd.DataFrame(data)

df['Month'] = df['Date'].dt.month_name()

df['Quarter'] = 'Q' + df['Date'].dt.quarter.astype(str)

df
```

```python
pd.pivot_table(df, values='Sales', index='Product' ,columns='Month')
```

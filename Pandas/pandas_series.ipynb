## Pandas Series

```python
import numpy as np
import pandas as pd
```

```python
labels = ['a', 'b', 'c']
list = [1, 2, 3]
arr = np.array([1, 2, 3])
d = {'a': 1, 'b': 2, 'c': 3}
```

```python
pd.Series(list)
```

```python
pd.Series(data=list, index=labels)
```

```python
pd.Series(arr)
```

```python
pd.Series(d)
```

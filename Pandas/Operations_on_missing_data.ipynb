## Operations on missing data

```python
import pandas as pd
import numpy as np
```

```python
data = {
    'A': [1, 2, np.nan, 4, 5, np.nan, 7, 8, 9, 10],
    'B': [np.nan, 2, 3, 4, 5, 6, np.nan, 8, 9, 10],
    'C': [1, 2, 3, np.nan, np.nan, 6, 7, np.nan, 9, 10],
    'D': [1, np.nan, np.nan, np.nan, 5, 6, 7, 8, np.nan, 10],
    'E': [np.nan, 12, 13, 14, np.nan, 16, 17, 18, 19, 20],
    'F': [21, 22, np.nan, 24, 25, np.nan, 27, 28, 29, 30],
    'G': [31, np.nan, 33, 34, 35, 36, np.nan, 38, 39, 40],
    'H': [41, 42, 43, np.nan, 45, 46, 47, np.nan, 49, 50]
}

df = pd.DataFrame(data)
df
```

Find the missing data

```python
df.isna()
```

```python
df.isna().sum(axis=1)
```

```python
df.isna().any()
```

Removing the missing data

```python
df.dropna()
```

```python
df.dropna(thresh=6)
```

Filling the missing data

```python
df
```

```python
df.fillna(0)
```

```python
df.fillna({'A': 0, 'B': 1, 'C': 2, 'D': 3, 'E': 4, 'F': 5, 'G': 6, 'H': 7})
```

```python
df.fillna(df.mean())
```

```python
from sqlalchemy import column
from streamlit import columns


data_list = [
    ['Alice', 25, 'New York', 50000],
    ['Bob', 30, 'Los Angeles', 60000],
    ['Charlie', 35, 'Chicago', 70000],
    ['David', 28, 'Houston', 55000],
    ['Emma', 24, 'Phoenix', 48000],
    ['Frank', 40, 'Philadelphia', 85000],
    ['Grace', 32, 'San Antonio', 62000],
    ['Henry', 29, 'San Diego', 58000],
    ['Isabella', 27, 'Dallas', 52000],
    ['Jack', 31, 'San Jose', 65000],
    ['Kevin', 45, 'Austin', 90000],
    ['Lily', 26, 'Jacksonville', 51000],
    ['Michael', 38, 'Columbus', 75000],
    ['Nina', 33, 'Charlotte', 68000],
    ['Oliver', 36, 'Seattle', 80000]
]
columns = ['name', 'age', 'city', 'salary']
df = pd.DataFrame(data=data_list, columns=columns)
df
```
